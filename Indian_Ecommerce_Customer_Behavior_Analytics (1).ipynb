```python
# Function to identify outliers using IQR

def detect_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return outliers, lower_bound, upper_bound
```

```python
# Detect unit price outliers

price_outliers, lower, upper = detect_outliers(
    df_clean,
    'unit_price'
)

print("Unit price lower bound:", lower)
print("Unit price upper bound:", upper)
print("Number of unit price outliers:", len(price_outliers))
```

```python
# Detect time-on-site outliers

time_outliers, lower, upper = detect_outliers(
    df_clean,
    'time_on_site_sec'
)

print("Time-on-site lower bound:", lower)
print("Time-on-site upper bound:", upper)
print("Number of time-on-site outliers:", len(time_outliers))
```

# Task 7: Business Insights & Recommendations

## Key Business Insights

### 1. Overall E-Commerce Performance
The dataset contains 25,000 customer visits and generated approximately ₹10.12 million in total revenue. This provides an overall view of the e-commerce business performance.

### 2. Conversion Rate
The overall conversion rate is 22.46%. Out of 25,000 customer visits, approximately 5.62K resulted in purchases, while 19.38K visits did not result in purchases.

### 3. Product Category Performance
Product Category Code 2 generated the highest revenue at approximately ₹2.04 million, followed by Product Category Code 6 at approximately ₹1.93 million.

### 4. Marketing Channel Performance
Marketing Channel Code 5 generated the highest revenue at approximately ₹1.82 million and recorded the highest number of purchases.

### 5. Device Performance
Device Type Code 1 contributed the highest revenue at approximately ₹5.01 million, representing about 49.49% of total revenue.

### 6. User Type Performance
User Type Code 1 generated higher revenue and showed a higher conversion rate than User Type Code 0, indicating stronger purchasing behavior from this user segment.

### 7. Cart Abandonment
A significant proportion of customer visits involved cart abandonment. This indicates an opportunity to improve the checkout process and encourage customers to complete their purchases.

## Business Recommendations

### 1. Improve Conversion Rate
The business should optimize the checkout process, reduce unnecessary steps, and provide clearer purchase information to convert more visitors into customers.

### 2. Reduce Cart Abandonment
The business should analyze why customers abandon their carts and consider strategies such as simplified checkout, reminders, and better product or pricing information.

### 3. Focus on High-Performing Channels
Marketing efforts can be strengthened on Marketing Channel Code 5 because it generated the highest revenue and purchase activity.

### 4. Promote High-Performing Categories
Products in Product Category Codes 2 and 6 should receive additional promotional attention because they generated the highest revenue.

### 5. Optimize the Shopping Experience
The business should monitor customer behavior across different device types and improve the shopping experience for devices with lower revenue or conversion performance.

## Conclusion

The analysis shows that product categories, marketing channels, device types, and customer behavior have an important relationship with e-commerce performance. The Power BI dashboard provides an interactive way to monitor revenue, purchases, customers, conversion rate, and purchasing behavior. These insights can help the business improve customer conversion, reduce cart abandonment, and focus resources on high-performing areas.
